```python
from pathlib import Path

import duckdb
import pandas as pd

pd.set_option("display.max_columns", 100)
pd.set_option("display.max_rows", 200)

CANDIDATE_DB_PATHS = [
    Path("../database/nutrition_data_filtered.duckdb"),
    Path("database/nutrition_data_filtered.duckdb"),
    Path("/home/harsh_wadhwaniai_org/eda-project/database/nutrition_data_filtered.duckdb"),
]

DB_PATH = next((p.resolve() for p in CANDIDATE_DB_PATHS if p.exists()), None)
if DB_PATH is None:
    raise FileNotFoundError(
        "Could not find nutrition_data_filtered.duckdb. Run scripts/build_nutrition_db.py first."
    )

con = duckdb.connect(str(DB_PATH), read_only=True)
print(f"Connected to: {DB_PATH}")
```

```python
tables_df = con.sql("SHOW TABLES").df()
display(tables_df)

if not tables_df.empty:
    table_name_col = tables_df.columns[0]
    table_counts = []
    for table_name in tables_df[table_name_col]:
        row_count = con.sql(f"SELECT COUNT(*) AS c FROM {table_name}").df().iloc[0]["c"]
        table_counts.append({"table": table_name, "rows": row_count})

    display(pd.DataFrame(table_counts).sort_values("rows", ascending=False))
```

```python
table_name = "nutrition_data"

print(f"Schema for: {table_name}")
display(con.sql(f"DESCRIBE {table_name}").df())

print("Sample rows")
display(con.sql(f"SELECT * FROM {table_name} LIMIT 10").df())
```

```python
cols_df = con.sql("PRAGMA table_info('nutrition_data')").df()
column_names = set(cols_df["name"].tolist())

print("Column metadata")
display(cols_df[["name", "type", "notnull"]])

print("Dataset size")
display(con.sql("SELECT COUNT(*) AS total_rows FROM nutrition_data").df())

if "District" in column_names:
    print("Top districts by record count")
    display(
        con.sql(
            """
            SELECT District, COUNT(*) AS records
            FROM nutrition_data
            GROUP BY District
            ORDER BY records DESC
            LIMIT 10
            """
        ).df()
    )

if "Date" in column_names:
    print("Date range")
    display(
        con.sql(
            """
            SELECT MIN(Date) AS min_date, MAX(Date) AS max_date
            FROM nutrition_data
            """
        ).df()
    )
```
